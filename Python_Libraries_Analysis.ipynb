```python
# import all libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
# Dataset loaded
df = pd.read_csv(
    "/content/Sample - Superstore.csv",
    encoding="cp1252"
)
```

```python
# Pandas sales and profit analysis
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Median Sales:", df["Sales"].median())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())

print("Total Profit:", df["Profit"].sum())
print("Average Profit:", df["Profit"].mean())
```

```python
# NumPy Mathematical calculations
sales = df["Sales"].to_numpy()

print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Maximum:", np.max(sales))
print("Minimum:", np.min(sales))
print("Standard Deviation:", np.std(sales))
print("Total Sales:", np.sum(sales))
```

```python
# Category-wise sales
df["Category"].value_counts()
```

```python
# Category-wise sales
category_sales = df.groupby("Category")["Sales"].sum()
category_sales
```

```python
# Matplotlib bar chart
category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
sales['Order Date'] = pd.to_datetime(sales['Order Date'])

daily_sales = sales.groupby('Order Date')['Sales'].sum()
```

```python
plt.figure(figsize=(10,4))

plt.plot(daily_sales.index, daily_sales.values)
plt.title("Daily Sales Revenue")
plt.xlabel("Order Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
cat_sale = sales.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8,5))

plt.pie(cat_sale.values, labels=cat_sale.index, autopct='%1.1f%%')

plt.show()
```

```python
category_sales = sales.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8,5))

plt.bar(category_sales.index, category_sales.values)
plt.title("Category Wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
